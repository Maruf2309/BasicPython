# Lambda Fuction

````python
# Lambda Function
result = lambda x: x*2
print(result)
````

**Why its returning object instead of actual result ?**
- This happens because:
    - Its printed function itself, not the result
    - Lambda Function create an anonymous function and store it in variable result, so the result is now, not actual result
- You must CALL the function: `print(result(2))`


🎯 Full Example

````python
# Lambda Function
result = lambda x: x * 2
print(result(3))
````

🚀 Bonus: Lambda with Multiple Arguments

````python
multiply = lambda x,y: x * y
print(multiply(3, 4))
````

💡 Where Lambda Is Commonly Used
```bash
map()
filter()
sorted()
pandas apply()

````python
numbers = [1, 2, 3]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
````

# Chapter 2: Basic OOP Version (Over Same Thing)

````python
# ✅ Step 1: Create a Class

class Calculator():
    def multiply(self, x):
        return x * 2

# ✅ Step 2: Create Object (Instance)
calc = Calculator()

# ✅ Step 2: Create Object (Instance)
result = calc.multiply(2)
print(result)
````

**🧠 Important Concept**

`self` means:
- The object(calc) calling the method
- it must be the first parameter in the class method

You don’t pass self manually — Python handles it.


🔥 Better Version (More Practical)

````python
# Instead of fixed *2, let’s multiply by any number.
class Calculator:
    
    def multiply(self, x, y):
        return x * y

# Call
calc = Calculator()
print(calc.multiply(3, 4))
````

🚀 More Advanced (Using Constructor)

````python
# Now let's store a default multiplier inside the object.
class Calculator:
    
    def __init__(self, multiplier):
        self.multiplier = multiplier
    
    def multiply(self, x):
        return x * self.multiplier

# Call
calc = Calculator(5)
print(calc.multiply(2))
````

🧠 What’s Happening Here?

- `__init__` runs when object is created
- `self.multiplier` stores value inside object

Every object can have different multiplier
